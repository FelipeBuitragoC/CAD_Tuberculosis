## 1 · 🔬 Extracción de Características

> **Definición:** Es el proceso mediante el cual se transforma información compleja, como una imagen,
> en un conjunto de variables numéricas con propiedades relevantes para el problema.

Es una fase **crucial** en la clasificación de enfermedades respiratorias: obtiene información de alto
nivel (color, forma, textura) asegurando que solo se consideren las áreas relevantes.

```
Radiografía  →  Extracción  →  Vector de características  →  Clasificador ML  →  Diagnóstico
```

📚 *Referencias: Tonni et al. (2025), Prince et al. (2025)*

## 2 · 📊 Características GLCM
### *Gray-Level Co-Occurrence Matrix*

Modelo estadístico que considera conexiones espaciales entre píxeles, capturando información
textural sobre la distribución de intensidades. Ayuda a capturar la complejidad de los
patrones en enfermedades respiratorias.

---

### 🔸 Contraste (`contrast`)

| Propiedad | Descripción |
|-----------|-------------|
| **Qué mide** | Diferencia de intensidad entre píxel de referencia y píxeles vecinos |
| **Valor bajo** | Textura uniforme, transiciones suaves |
| **Valor alto** | Variaciones significativas → posibles anomalías en tejido pulmonar |

---

### 🔸 Disimilitud (`dissimilarity`)

| Propiedad | Descripción |
|-----------|-------------|
| **Qué mide** | Diferencia absoluta entre niveles de gris de pares de píxeles |
| **vs Contraste** | Usa diferencia absoluta, **no** eleva al cuadrado |

---

### 🔸 Homogeneidad (`homogeneity`)

| Propiedad | Descripción |
|-----------|-------------|
| **Qué mide** | Concentración de elementos de la GLCM alrededor de su diagonal |
| **Valor alto** | Niveles de gris similares · menor contraste |
| **Pulmón sano** | Valores más altos de homogeneidad |

---

### 🔸 Energía (`energy`)

| Propiedad | Descripción |
|-----------|-------------|
| **Qué mide** | Uniformidad local de niveles de gris |
| **Valor alto** | Mayor similitud entre píxeles · pulmón sano |

---

### 🔸 Correlación (`correlation`)

| Propiedad | Descripción |
|-----------|-------------|
| **Qué mide** | Relación lineal entre valores de niveles de gris en la GLCM |

## 3 · 📐 Características Basadas en la Forma
### *Shape-based Features*

Capturan características **geométricas** esenciales para la clasificación de enfermedades.
Permiten identificar lesiones pulmonares irregulares.
*(Ayaz et al. 2021, Tonni et al. 2025)*

---

### 🔹 Perímetro (`perimeter`)
Contorno de la lesión pulmonar que indica su extensión.
⚠️ Un **perímetro aumentado** puede indicar presencia de infección.

---

### 🔹 Área (`area`)
Tamaño de la región pulmonar: número de píxeles dentro de un contorno específico.

---

### 🔹 Circularidad (`circularity`)
Mide la **redondez** del área lesionada.
Desviaciones de la circularidad esperada indican anomalías.

---

### 🔹 Extensión (`extension`)
Captura la distribución espacial de anomalías pulmonares:
- Resalta bordes, curvas y extremos
- Extrae información mediante formas
- Codifica la información del píxel central

---

### 🔹 Solidez (`solidity`)
**Compacidad** de una lesión pulmonar.

```
Solidez = Área de la lesión / Área de su casco convexo
```

Una solidez baja indica bordes irregulares o huecos internos.

<a id="lbp"></a>

## 4 · 🔲 Características LBP
### *Local Binary Pattern*

Modelo estadístico eficiente para extraer características de imágenes.
Los valores LBP se obtienen **binarizando** las diferencias entre píxeles adyacentes.

```
Vecindad de píxeles  →  Comparación con píxel central  →  Código binario LBP
```

---

| Feature | Descripción |
|---------|-------------|
| **ConvexArea** | Área del casco convexo que envuelve la región de interés |
| **lbp_energy** | Uniformidad de los patrones LBP. Valores altos → textura regular |
| **lbp_entropy** | Complejidad de los patrones LBP. Valores altos → mayor heterogeneidad |

<a id="pulmonar"></a>

## 5 · 🫁 Características Pulmonares
### *Lung Features*

Se evalúa la **densidad pulmonar** comparando zonas superior, media e inferior de ambos pulmones.
Se calcula blancura anormal (mayor densidad) o negrura anormal (menor densidad).

---

### 🫁 Densidad Media (`mean_density`)

| Valor | Significado |
|-------|-------------|
| **Alto** | Mayor suma acumulativa de niveles de gris |
| **Bajo** | Menor densidad pulmonar |

---

### 🫁 Densidad Estándar (`std_density`)
Mide la **variabilidad** en los valores de intensidad de los píxeles respecto al valor medio.

---

### 🫁 Volumen Pulmonar (`lung_volume`)

| Tipo | Uso clínico |
|------|-------------|
| **Dinámico** | Diagnóstico de enfermedades obstructivas |
| **Estático** | Evalúa defectos obstructivos y restrictivos |

⚠️ Volumen anormal puede indicar: cardiomegalia, neumonía viral o bronquitis.

---

### 🫁 Puntos Clave SIFT
#### *Scale-Invariant Feature Transform*

Extrae características locales mediante **coordenadas invariantes a la escala**.

- ✅ Filtra patrones irregulares y de bajo contraste
- ✅ Resistente a deformaciones locales y errores de detección

## 6 · 🌀 Características DC-GLM
### *Dynamic Co-Occurrence Grey Level Matrix*

Medidas numéricas de textura (contraste, homogeneidad, energía, correlación) que convierten
información visual en variables de entrada para algoritmos de aprendizaje automático.

---

### 🌀 Momento Angular Secundario (`ASM` — Angular Second Moment)

| Valor | Interpretación |
|-------|----------------|
| **Alto** | Distribución de textura más **uniforme** |
| **Bajo** | Textura más heterogénea o compleja |

> 💡 El ASM es equivalente a la energía GLCM estándar, aplicado en el contexto dinámico del DC-GLM.

<a id="otros"></a>

## 7 · 📖 Otros Términos Clave

---

### 📌 Coocurrencia
Concepto estadístico que describe la **aparición conjunta** de dos valores dentro de
una determinada relación espacial en una imagen.

---

### 📌 Filtro de Gabor (*Gabor Filter*)
Operador para detectar estructuras con determinadas **frecuencias y orientaciones**.
Combina una función **gaussiana** con una **señal sinusoidal**.
Útil para identificar patrones de textura y estructuras orientadas.

---

### 📌 Filtrado Multiescala (*Multiscale Filtering*)
Analiza una imagen usando **diferentes escalas** simultáneamente:

| Escala | Detecta |
|--------|---------|
| Pequeña | Estructuras finas, detalles |
| Mediana | Regiones intermedias |
| Grande | Estructuras globales |

---

### 📌 Difusión Anisotrópica (*Anisotropic Diffusion*)
Reduce **ruido** sin eliminar bordes importantes.
La difusión depende de las características locales, preservando estructuras relevantes.

---

### 📌 Difusión Adaptativa Contextual (*Contextual Adaptive Diffusion*)
Método de **CAMSGNeT** que controla la difusión por región considerando:
- El **gradiente** de la imagen
- La **varianza local** de las intensidades

---

### 📌 Gradiente de Imagen (*Image Gradient*)
Representación matemática de **cómo cambia la intensidad** espacialmente.

| Gradiente | Interpretación |
|-----------|----------------|
| **Alto** | Transición fuerte → borde |
| **Bajo** | Región uniforme |
